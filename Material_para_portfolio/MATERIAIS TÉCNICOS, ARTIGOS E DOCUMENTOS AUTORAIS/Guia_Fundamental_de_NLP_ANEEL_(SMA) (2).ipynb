# 1. Conceitos gerais

## **O que é Inteligência Artificial (IA)?**
### _Fique por dentro de uma das tecnologias mais disruptivas atualmente_

<br>

Ainda hoje, é difícil definir o que é a **inteligência**. Porém, sabe-se que algumas tarefas exigem **inteligência humana** para serem realizadas (como cálculo, previsões, identificação de objetos, etc.). A grosso modo, quando nos deparamos com um sistema computacional que age como um ser humano (simular pensamentos humanos ou realizar atividades que exijam inteligência humana), dizemos que estamos lidando com uma **inteligência artificial**.

<br>

<div align="center">

![Inteligência Artificial](https://www.telesintese.com.br/wp-content/uploads/2023/04/Inteligencia-Artificial-IA-AI-freepik-2023.jpg)

</div>

<br>

A Inteligência Artificial é a fascinante área da ciência da computação que se dedica a criar sistemas capazes de realizar tarefas que normalmente exigiriam inteligência humana. Inspirada pela habilidade humana de aprender, raciocinar e resolver problemas, a IA busca dotar máquinas com a capacidade de imitar e, em muitos casos, superar as habilidades cognitivas humanas.

Para o campo da IA, um dos principais fundamentos é o conceito de "aprendizado": os sistemas são projetados para **extrair padrões** a partir de dados, ajustar seu desempenho com o tempo e usar esse conhecimento para resolver problemas. Assim como uma criança aprende a identificar um gato após ver várias imagens, os modelos de IA aprendem a reconhecer padrões complexos em grandes conjuntos de dados, que chamaremos de _datasets_ neste guia.

A promessa da IA é a automação de tarefas complexas, a resolução de problemas desafiadores e o desenvolvimento de sistemas que compreendem, interagem e aprendem com o mundo ao seu redor.

É seguro dizer que, nos dias de hoje, a Inteligência Artificial é um dos campos mais revolucionários da tecnologia. Podemos notar aplicações da IA em produtos domésticos, na medicina, em grandes empresas, em corretoras financeiras, e em muitas outras áreas. Sendo assim, o entendimento de conceitos básicos de IA faz-se necessário para toda nossa equipe, que não só entenderá melhor os aspectos do projeto em que estamos trabalhando, mas também da tecnologia que nos cerca.

## **O que é Processamento de Linguagem Natural (NLP)?**
### _Entenda como o computador entende a comunicação humana_

<br>

<div align="center">

![NLP](https://images.ctfassets.net/63bmaubptoky/nygs3El6gGPEUAo5Jp-oU1itL4CIJQjqyA1IAIO95bY/fdf70442803105d877150202331c6a85/Header-What-is-Natural-Language-Processing-UK-CA-Header.png)

</div>

<br>

Antes de nos aprofundarmos em NLP, precisamos saber o que é **linguagem natural**. A linguagem natural é a forma em que os seres humanos se comunicam, seja ela de forma escrita ou falada. Essa organização em palavras, com tons e fonética próprios, definem a linguagem natural.

A linguagem natural possui diversas regras que organizam as frases de forma que estas tenham um significado. O problema é que essas regras são complexas, podendo mudar de acordo com o idioma, o tempo verbal, etc. Hoje em dia, em relação à linguagem natural, o desafio da tecnologia é fazer com que:

<br>

1. O computador entenda a sintaxe (forma de organizar as palavras) e a semântica (significado) das frases;
2. Fazer com que o computador use o conteúdo linguístico para realizar uma **tarefa específica** de interesse ao ser humano que opera o computador;
3. A tarefa executada deve ser feita de forma **precisa e confiável**, sem alucinações.

<br>

Porém, um computador não é capaz de entender tal complexidade naturalmente. A única linguagem entendida por computadores é a chamada **linguagem de máquina**, uma linguagem de **baixo nível**, que se resume em sequências de "zeros e uns" para indicar comandos.

Dessa forma, o **Processamento de Linguagem Natural** é uma subdivisão do campo de IA, que consiste em estabelecer uma ponte entre a linguagem natural (escrita ou falada), e as máquinas (que apenas interpretam linguagem de máquina). Com isso, a interação humano-computador é muito facilitada, tornando possível que o computador realize diversas tarefas complexas de interesse da humanidade.

Para exemplificar um produto atual que utiliza NLP, podemos pensar nos assistentes inteligentes, como a _Alexa_, da Amazon. O usuário pode dizer (linguagem falada), por exemplo, "Alexa, adicione 'Tomate' à minha lista de compras". A Alexa então recebe a mensagem, extrai significado dessa mensagem e realiza a ação de adicionar o item "Tomate" no aplicativo de lista de compras.

## **O que é um modelo de IA (em especial, um LLM)?**
### _Veja como esses modelos podem facilitar nossa vida_

<br>

Um modelo de IA é um programa que foi treinado em um conjunto de dados (dataset) para realizar tarefas de inteligência artificial. Um modelo é composto por uma ou mais redes neurais artificiais (explicadas na próxima seção) e funciona da seguinte forma:

1. Recebe uma entrada (input ou prompt) ou um conjunto de entradas;
2. Processa essa entrada internamente;
3. Dá uma resposta para essa entrada. Essa resposta é baseada no treinamento que o modelo teve.

Diariamente, a comunidade de IA desenvolve diversos modelos. Existem modelos para realizar tarefas de NLP (vide Alexa), outros para analisar conjuntos de dados e chegar em previsões de preços de objetos para anos futuros, outros para fazer com que o computador reconheça imagens (vide sistemas de reconhecimento facial) e assim por diante.

Assim podemos, através das linguagens de programação, acessar e modificar esses modelos para atenderem aos nossos problemas específicos.

**Mas por que utilizamos modelos de IA?**

Imagine que você quer ensinar uma criança a andar de bicicleta. Se você fizesse isso como um programador (com algoritmos normais), precisaria dizer coisas como: "coloque seus pés nos pedais. Em seguida, aplique uma força de 120N para baixo em uma perna, enquanto relaxa a outra. Se pender para a direita, contrabalanceie o peso para a esquerda. Caso contrário...".

Perceba como uma tarefa complexa, se ensinada no formato de algoritmos, pode ser muito difícil de ser programada!

Os modelos de IA seriam como a criança que sabe que deve aprender a andar de bicicleta e não cair. Para isso, pedala de um jeito X, cai, testa um novo jeito Y, cai novamente, e segue experimentando até aprender que se pedalar primeiro de jeito Y e depois de jeito X, consegue se manter na bicicleta por mais tempo, ou seja, aprende um padrão que funciona para resolver esse problema. Assim, a IA aprende os padrões de forma autônoma, finalmente se tornando capaz de realizar essa tarefa complexa, facilitando (e muito) a vida dos programadores e da sociedade.

**E o que é um LLM?**

<br>

<div align="center">

![Neuronio](https://datascientest.com/en/wp-content/uploads/sites/9/2023/06/large-language-models-1024x512.png)

</div>

<br>

Um LLM (Large Language Model ou Grande Modelo de Linguagem) é um tipo avançado de modelo de IA que busca entender, gerar e interagir com a linguagem natural humana. Um LLM é treinado com muitos textos tirados de artigos, sites, etc. **Esse tipo de modelo se torna especializado em realizar tarefas relacionadas à linguagem (tarefas de NLP)**. Cabe ressaltar que os LLMs são compostos por redes neurais artificiais profundas, que serão explicadas neste guia.

No projeto que estamos trabalhando, estaremos usando um LLM, ou seja, um grande modelo de NLP. Nas próximas sessões, abordaremos novas informações sobre esse tipo de modelo.

## **O que é uma Rede Neural Artificial?**
### _Saiba como são os modelos de IA por trás dos panos_

<br>

#### **O conceito geral**


O cérebro humano possui cerca de 86 bilhões de neurônios e mais de 100 trilhões de conexões entre eles. Ou seja, esses neurônios estão conectados entre si, formando uma grande rede de neurônios, que chamamos de rede neural.

Essa estrutura é responsável pela interação entre nós, seres humanos, e o ambiente ao nosso redor. Através de nossas redes neurais, realizamos nossas funções vitais (respiração, batimentos cardíacos e etc.) e respondemos à estímulos ambientais que recebemos em nossos órgãos sensoriais.

Na tentativa de alcançar a capacidade humana, a tecnologia computacional se desenvolveu muito. Hoje em dia, a Inteligência Artificial é baseada no funcionamento do nosso cérebro. Dessa forma, foi desenvolvido o que chamamos de **rede neural artificial** (RNA). Como o nome sugere, é uma imitação das redes de neurônios presentes em nosso cérebro.

#### **Mas como nós, seres humanos, aprendemos?**

Para aprendermos como funcionam as redes neurais artificiais e as IAs como um todo, precisamos entender um pouco de neurociência.

Ao ler este material, você está passando pelo processo de aprendizagem. Vamos utilizar este fato para ilustrar o que ocorre (e está ocorrendo nesse momento) no seu cérebro!

> Faça um teste e tente responder essa pergunta neste exato momento: _**como o ser humano aprende uma informação?**_

Dividiremos essa etapa em 4 partes, que seguirão o seguinte fluxo:


<div align="center">
<br>

 `Recepção do estímulo -> Processamento do estímulo -> Armazenamento da informação -> Resposta ao estímulo`

<br>

**_1 - Recepção do estímulo_**


Quando assistimos uma aula, ou lemos um guia como este, estamos recebendo estímulos do nosso ambiente (um professor falando, um livro, um guia na internet, etc.). Esses estímulos chegam ao nosso sistema nervoso através dos nossos órgãos sensoriais, como os olhos, os ouvidos e etc.

Nosso sistema nervoso periférico possui diversos neurônios, que recebem esses estímulos e os repassam quimicamente através de sinapses, até que cheguem ao sistema nervoso central, composto pelo encéfalo (que possui o cérebro) e pela medula espinhal.

<br>

<br>

**_2 - Processamento do estímulo_**


Chegando ao encéfalo, a informação passa pelo importantíssimo mecanismo de processamento.

Primeiramente, nossa **atenção** seleciona quais partes do estímulo possuem mais importância para o aprendizado. Ou seja, sua atenção está, nesse momento, selecionando as **palavras-chave** deste texto que vão te ajudar a entender o estímulo. Em outras palavras, a atenção atribui um **grau de importância**, ou seja, **um peso** para cada parte do estímulo.

Dessa forma, as partes mais importantes do estímulo possuem um **grau de importância** maior, e as menos importantes possuem um **grau de importância** menor.

Precisamos, neste ponto, lembrar que os estímulos são apenas sinais químicos, e ainda não possuem significado para o cérebro.

Para resolver isso, o cérebro utiliza um mecanismo de análise que busca **extrair significado** das partes mais importantes do estímulo (enquanto você lê, seu cérebro está extraindo significado deste texto, dando mais atenção às palavras e trechos mais importantes), transformando o estímulo em uma informação. Isso ocorre com a ativação dos neurônios.

_**Para que você entenda:**_ você possui muitos neurônios, certo? Quando você pensa na palavra "gato", por exemplo, alguns desses neurônios são ativados. Essa combinação de neurônios ativados é um padrão de ativação que simboliza a ideia do que é um gato. Quando você pensa na palavra "cachorro", outros neurônios são ativados, formando uma combinação diferente para simbolizar a ideia do que é um cachorro.

Na imagem abaixo, o bloco em cinza ilustra uma rede neural. Cada círculo dentro do bloco é um neurônio. Eles estão interconectados, formando uma rede neural. Os neurônios em preto estão desativados, enquanto os neurônios em vermelho estão ativados. Esse mosaico de ativação de neurônios é **chamado de padrão de ativação neural**. Cada pessoa ativa padrões diferentes para a mesma ideia, ou seja, você tem um padrão para a ideia do que é um gato, e seu pai tem um padrão diferente para a mesma ideia do que é um gato, por exemplo.

<br>
<div align="center">

</div>
<br>

Logo, ao dizermos que transformamos um estímulo em informação, estamos na verdade aprendendo novas combinações de ativação de neurônios, que simbolizam a informação desse novo estímulo.

Essas informações serão armazenadas na memória.

<br>

**_3 - Armazenamento da informação_**


O mecanismo de memória é responsável por armazenar esses padrões de ativação de neurônio. Assim, nosso cérebro é capaz de lembrar qual combinação de neurônios deve ser ativada para resgatar uma ideia.

Você já percebeu que você se esquece de informações que não usa no dia a dia? Por exemplo: suponha que você estudou sobre Processamento de Linguagem Natural. Se você não revisita esses conceitos com certa frequência, você tende a esquecer rapidamente dessas informações.

Quando começamos a esquecer sobre certa informação, significa que o cérebro está considerando essa informação como sendo menos importante, pois ela está em desuso. De certa forma, o cérebro está diminuindo os **graus de importância** dessa informação. Para resolver isso, precisamos revisar, não é?

Quando fazemos isso, estamos recalibrando os **graus de importância** das partes dessa informação, ou seja, estamos dizendo ao cérebro que é importante manter aquela informação em memória.

Assim, acabamos reforçando a capacidade de ativar a combinação de neurônios que remete a essa informação. Assim, quanto mais **treinamos** nosso cérebro para interagir com certa informação, mais fácil fica para utilizarmos corretamente esse conhecimento posteriormente.

<br>
<div align="center">
  <img src="https://iili.io/JXOednS.png">
</div>
<br>

>Os **graus de importância** dados pela atenção são muito importantes. Quando você estuda uma informação pela primeira vez, indiscutivelmente algumas coisas passam despercebidas. Às vezes, uma informação de um texto era relevante e você não percebeu. Você concorda que os graus de importância, nesse caso, **não** foram calibrados da melhor forma? Isso impacta na ativação da combinação de neurônios para essa informação.
>
>Por isso, quanto mais revisarmos certo material de estudo, maior a chance de calibrarmos melhor esses graus de importância, resgatando melhor essa informação na memória, posteriormente. Em outras palavras, se treinamos apenas uma vez, os **graus de importância** podem não estar tão bem ajustados. Quanto mais **treinamos**, melhor ajustamos nossos graus de importância para certa informação. E quanto mais calibrados os **graus de importância**, melhor será seu desempenho para resgatar essa informação.

Depois de tudo isso, ao responder a pergunta, você utilizou sua memória para ativar novamente esses neurônios e resgatar essas informações, uma por uma, para então formar uma resposta adequada à pergunta.

A pergunta foi um estímulo para você. Quando resgatamos essas informações na memória e "jogamos para fora" por meio da comunicação (escrita ou falada) ou por alguma ação, estamos dando uma resposta para esse estímulo.

#### **E como funcionam as redes neurais artificiais?**

Como dito anteriormente, as redes neurais artificiais (RNAs) são compostas por muitos neurônios (também artificiais), conectados entre si. Essas redes possibilitam que o computador realize atividades de Inteligência Artificial.

Você perceberá muitas semelhanças e diferenças entre as redes neurais humanas e as RNA. As semelhanças ocorrem pois as RNAs são baseadas no funcionamento das redes neurais biológicas. Já as diferenças ocorrem por conta das limitações que um sistema computacional possui em relação a sistemas biológicos.

Uma rede neural artificial utiliza seus neurônios artificiais para processar um estímulo. Depois de processar esse estímulo, a rede gera uma resposta a esse estímulo, assim como os seres humanos.

Comumente, quando estamos lidando com RNAs, chamamos esse estímulo de entrada. Por isso, usaremos essa terminologia neste guia.

**Mas como isso acontece?**

Para explicar o funcionamento de uma RNA, vamos dar um exemplo relacionado a NLP.

Vamos supor que você abriu o _ChatGPT_. Em seguida, você envia ao modelo o seguinte texto: **"Bom dia! Eu gostaria de tirar uma dúvida: qual a capital do Japão?"**. Naturalmente, você espera que ele processe essa pergunta e te dê uma resposta adequada, certo?

<br>

Para chegar à essa resposta, o _ChatGPT_ utiliza suas redes neurais. Simplificadamente, funciona da seguinte forma:

<br>

**1) Pré-processamento**

Nesse exemplo, o texto que você enviou ao _ChatGPT_ é a entrada (estímulo).

Antes de tudo, há um problema. As redes neurais precisam que o estímulo esteja em um formato correto para entrar na rede. Ou seja, para que o estímulo passe pela rede neural artificial, o modelo precisa **preparar** esse estímulo para um formato que a rede neural entenda.

Em modelos de NLP, como o do _ChatGPT_, as redes neurais precisarão quebrar o texto de entrada em pequenas partes.

O processo de se "quebrar" o texto em pequenas partes é chamado de **tokenização**, que será explicada em mais detalhes nas próximas seções. Depois da quebra, cada pequena parte da entrada é chamada de **token**.

Para esse exemplo, vamos supor que o texto tenha sido quebrado palavra por palavra, conforme a imagem abaixo:

<br>
<div align="center">
  <img src="https://iili.io/JWQnnXp.png">
</div>
<br>

Depois de quebrarmos o texto em pequenas partes (tokens), ainda temos um pequeno problema: as redes neurais funcionam realizando **cálculos matemáticos**. Ou seja, o modelo precisa **transformar essas partes em números!**

Chamamos essa conversão de token para número de **codificação**. Cada token é **convertido em um número identificador (ID) que o representa**. Esse número, mesmo que de forma indireta, será utilizado nos cálculos da rede neural.

Não se preocupe com o trecho "de forma indireta". Você entenderá mais sobre tokenização, codificação e outros processos nas próximas seções. Por enquanto, o mais importante a se entender é: **a entrada é quebrada em pequenos tokens, cada token é convertido em um ID que será usado nos cálculos dos neurônios**.

Para entender melhor essa ideia, verifique a imagem abaixo:

<br>

</div>
<br>


Depois que a entrada foi preparada para adentrar a rede neural, esse estímulo tokenizado e codificado em IDs é enviado para a rede.

<br>

**2) Grau de importância**

Você concorda que, no texto, existem sequências de palavras que são mais importantes do que outras para dar uma resposta a uma pergunta?

Observe novamente o nosso exemplo:

<br>

Baseado na resposta do ChatGPT para o nosso exemplo, podemos dizer que ele considerou:

1. **Palavras da sequência "Bom dia!"**: possuem certa importância, pois o modelo gerou outra saudação de "bom dia" em sua resposta;
2. **Palavras da sequência "Eu gostaria de tirar uma dúvida:"**: possuem menor importância, pois o modelo poderia desprezar essa parte da nossa entrada para chegar à mesma resposta que ele gerou;
3. **Palavras da sequência "qual a capital do Japão?"**: possuem maior importância, pois carregam a solicitação que fizemos ao modelo. Sem essas palavras, o modelo não saberia a pergunta que deveria responder.


Você se lembra do **mecanismo de atenção** das redes neurais humanas (que dava graus de importância)? Analogamente, os neurônios das redes neurais atribuem **pesos (ou graus de importância)** para cada parte da entrada, conforme a imagem abaixo. Esses pesos, na rede neural artificial, são pequenos números que quantificam a importância de cada parte da entrada.

Dessa forma, as partes que possuem maior peso são consideradas mais importantes para chegarmos à resposta. Já as de menor peso são consideradas menos importantes.

**Mas como os pesos são atribuídos?**

Mais detalhes sobre isso serão explorados nas próximas seções. Mas podemos dar uma prévia nesse momento:

Inicialmente, os pesos atribuídos são **aleatórios**. Depois, com o treinamento do modelo, os pesos são ajustados de acordo com o dados de treino. Você entenderá mais sobre isso nas próximas seções.

<br>

**3) Processamento**

Vimos então que as palavras são enviadas para os neurônios que existem dentro da rede neural artificial.

Você lembra que uma rede neural artificial é um conjunto de neurônios artificiais, certo? Além disso, esses neurônios estão conectados entre si e, em conjunto, processam um estímulo e geram uma resposta.

Porém, a estrutura de uma rede neural artificial é um pouco mais complexa do que nossas ilustrações anteriores. Na verdade, temos várias camadas de neurônios que se seguem até o final da rede, como ilustra a imagem a seguir:

<br>

</div>

<br>

As redes neurais artificiais são organizadas em camadas de neurônios, como na imagem acima. Cada tipo de camada realiza uma função diferente. Em resumo, temos 3 tipos de camadas de neurônios em uma rede neural artificial:

1. **Camada de entrada**: recebe o estímulo dividido em partes;
2. **Camadas ocultas**: são responsáveis pelo processamento do estímulo;
3. **Camada de saída**: recebe o estímulo processado pelas camadas ocultas e gera uma resposta adequada ao estímulo.

Sendo assim, no nosso exemplo, algo mais assertivo poderia ser ilustrado da seguinte forma:

<br>

</div>
<br>

Os primeiros neurônios da rede funcionam atribuindo pesos para cada parte da entrada e realizando **cálculos**. Esses cálculos utilizam as **partes da entrada juntamente com seus pesos**. Ou seja, eles recebem esses elementos como entrada e, depois de realizar um cálculo interno, geram um resultado numérico.

Essa resposta numérica é passada como entrada para o próximo neurônio da rede, que vai receber essa resposta, atribuir um peso para essa resposta, e realizar outro cálculo interno. O resultado desse novo cálculo é passado para o próximo neurônio e assim por diante. Você percebe que, de certa forma, o estímulo está passando pelos neurônios do início ao fim da rede? Esse comportamento é chamado de **propagação**.

A importância desses cálculos será elucidada posteriormente.

>Perceba: as respostas numéricas de cada neurônio vão depender das entradas e dos pesos, certo? Podemos ter respostas erradas nos neurônios se as palavras da entrada estiverem no idioma incorreto, mal escritas, ou se os pesos (graus de atenção) não estiverem bem ajustados.
>
>Parece abstrato, mas ocorre o mesmo com nós, seres humanos: se voce recebe, no seu _Whatsapp_, a mensagem "Здравствуйте, я работаю в ANEEL", o entendimento é fácil ou difícil? Se você está com dor de barriga durante uma palestra, seu mecanismo de atenção vai desviar sua atenção para a dor de barriga. Vai ser mais fácil ou difícil absorver o conteúdo da palestra? Percebe como as palavras bem escritas no idioma correto e a atenção bem direcionada podem fazer a diferença?
>
> Isso ocorre porque você foi treinado para interagir com as linguagens que você praticou desde criança, ou com um curso de línguas. Além disso, mesmo com o idioma e escrita corretos, você não entenderá muita coisa se sua atenção for desviada com muita frequência.
>
> Com os modelos de NLP, acontece a mesma coisa. Eles são treinados com dados que às vezes são, em sua maioria, em inglês. Isso faz com que uma entrada em mandarim seja mal interpretada pelo modelo, pois ele não treinou pra isso! Para responder bem em mandarim, você pode treinar o modelo com dados em mandarim e, assim, ajustar seus **pesos** de acordo com essa nova fonte de dados.
>
>Sendo assim, as respostas da rede neural podem ser melhores ou piores, a depender da qualidade da entrada e, principalmente, dos pesos atribuídos a cada parte da entrada.

<br>

Pois bem. Agora você pode (e deve) estar se perguntando: **qual o motivo desses cálculos? Como eles ajudam o modelo a chegar na resposta correta?** Desvendaremos isso agora.

Um modelo de NLP, por mais que possua redes neurais que agem de forma semelhante ao pensamento humano, ainda possui (muitas) limitações. Enquanto nós, seres humanos, geramos respostas à estímulos por ativação de padrões neuronais complexos, as redes neurais artificiais utilizam a matemática das probabilidades.

Na camada de saída da rede, os cálculos que vimos anteriormente, realizados pelos neurônios artificiais, serão transformados em probabilidades que fazem sentido para nós, seres humanos.

Na resposta, se o modelo der uma probabilidade muito alta para uma palavra, isso indica que a rede está "dizendo" que tem mais certeza de que essa palavra será uma boa escolha para se colocar na resposta. Em outras palavras, essas probabilidades indicam o grau de certeza que o modelo tem em dar certa palavra como resposta.

Em um modelo moderno de NLP, como o _ChatGPT_, essas probabilidades são geradas da seguinte forma:

1. Os modelos conhecem muitas palavras. O ChatGPT (versão 3.5 Turbo) conhece cerca de 470 mil palavras diferentes. O conjunto de palavras que um modelo conhece recebe o nome de vocabulário;
2. Os neurônios calculam números baseados nas entradas e nos seus pesos, como vimos acima;
3. Essas respostas numéricas baseadas no estímulo são propagadas até a camada de saída;
3. Na camada de saída, essas respostas numéricas são usadas para gerar uma probabilidade para cada palavra do vocabulário do modelo. Essa é a probabilidade de cada palavra se "encaixar" no contexto do texto de entrada;
4. A palavra que tiver a maior probabilidade de ser "encaixada" no contexto da nossa entrada será escolhida para ser gerada na resposta.

<br>
<div align="center">
  <img src="https://iili.io/JWtxZkF.png">
</div>
<br>

Você concorda que, para responder corretamente nossa frase **"Bom dia! Eu gostaria de tirar uma dúvida: qual a capital do Japão?"**, a probabilidade de se gerar a palavra "Bom" (para responder ao "bom dia" da nossa frase) deve ser bem maior que a probabilidade de se iniciar a resposta com "Gambá"? Pois é. É assim que um modelo moderno de NLP trabalha.


Ok, geramos a palavra "Bom" usando probabilidades que saíram dos cálculos dos neurônios. E agora, como o modelo constrói o resto da resposta? É simples: ele roda novamente seu algoritmo para gerar a próxima palavra. Porém aqui ocorre um passo muito importante para modelos de NLP: **a próxima palavra gerada não será baseada apenas no nosso texto de entrada, mas também na resposta que o modelo já gerou!**

Observe as imagens a seguir para analisar esse processo:

Ou seja, no nosso exemplo, para gerar a segunda palavra da resposta, o _ChatGPT_ usou o mesmo texto de entrada juntamente com a palavra "Bom" (já gerada) para gerar "dia!". Depois, utilizou nosso texto de entrada mais uma vez, mas agora em conjunto com a resposta já gerada "Bom dia!", para gerar a palavra "A". O modelo segue fazendo isso até completar a resposta.

Entenda: se o modelo fosse gerar a resposta palavra por palavra usando **apenas a mesma frase de entrada, sem considerar a resposta já gerada**, estaria apenas repetindo o mesmo algoritmo, sem mudar nada na entrada. Sendo assim, ele poderia gerar a palavra "Bom" repetidas vezes, pois se nada muda na entrada, é razoável que nada mude na saída, certo? Então a resposta gerada, nesse caso, poderia ser: "Bom Bom Bom (...)". Nada "Bom", né?

Além disso, superficialmente, elucidamos um exemplo explicando como um modelo de NLP age para gerar uma resposta através de uma entrada (estímulo).

## **Parâmetro? Hiperparâmetro? Qual a diferença?**
### _Aprenda de uma vez, sem confusão!_

<br>

Anteriormente, vimos que as redes neurais possuem diversos neurônios artificiais. Esses, por sua vez, atribuem **pesos** às suas entradas. Esses pesos têm a capacidade de determinar as respostas da rede neural. Quando treinamos uma rede neural, estamos ajustando esses pesos (ou graus de atenção) para gerar respostas cada vez mais assertivas.

Quando criamos uma IA baseada em RNAs, os pesos iniciais são **valores aleatórios**. Durante o treinamento de uma IA, os pesos são **ajustados** para atender o problema de interesse.

Dessa forma, podemos definir que os **parâmetros** são os componentes da rede neural que serão modificados **durante** o treinamento da IA, como os pesos.


Porém, antes do treinamento, precisamos definir **como o modelo treinará**. A quantidade camadas ocultas, o número de neurônios nas camadas ocultas e o número de vezes que a IA irá treinar em um mesmo conjunto de dados (epochs) são exemplos de configurações que serão definidas **antes** do treinamento.

A esses elementos que são definidos **antes do treinamento**, damos o nome de **hiperparâmetros**.

Concluindo, no treinamento de uma IA, definimos a estrutura do treinamento com os **hiperparâmetros**. Esse treinamento modificará os **parâmetros** da rede neural.

<br>

<div align="center">

Hiperparâmetro | Parâmetro
--- | ---
É definido **antes** do treinamento | É modificado **durante** o treinamento
Quantidade de neurônios, _epochs_, número de camadas ocultas (...) | Pesos

</div>

## **O que é "treinar uma IA"?**
### _Saiba mais sobre mecanismo chave dos modelos de IA_

<br>

Quando estamos trabalhando com modelos de IA, como os LLMs, precisamos realizar o treinamento desses modelos pelo seguinte motivo: **geralmente, os parâmetros (pesos) iniciais da rede neural são valores aleatórios**! Isso significa que a rede neural, quando criada, quase sempre dará resultados errados. Por isso, precisamos treinar a rede.

Uma vez que os parâmetros podem definir a ativação dos neurônios, é preciso que esses parâmetros sejam ajustados ao problema que queremos resolver. Em outras palavras, precisamos fazer com que os pesos sejam modificados, de forma que as respostas numéricas dos neurônios levem à probabilidades corretas na camada de saída, para assim gerar respostas corretas.

<div align="center">

<br>

![Treinamento de IA](https://images.ctfassets.net/3viuren4us1n/2CsVyr5yjxEiSsNVcMmB7E/480b044a46151c345ee8430270a77237/How-do-you-train-AI.jpg)

<br>

</div>

O treinamento ocorre baseado em um conjunto de dados, que chamamos de **dataset**. Analogamente, você pode imaginar uma tabela do Excel com linhas e colunas.

Exemplo de dataset usado para treinar alguns modelos de NLP:

<br>

</div>

<br>


Para que ocorra um treinamento, devemos decidir se esse será supervisionado (rotulado) ou não. **Em treinamentos supervisionados**, utilizamos datasets que contenham dados com **rótulos** (etiquetas que servem como gabaritos) para a IA entender qual o resultado esperado para aquela entrada.

Na prática, isso quer dizer que estamos dizendo ao modelo: "estou te entregando esse conjunto de dados. Nele, possuímos entradas e respostas para essas entradas. Aprenda que, se alguém utilizar essas entradas, eu quero que você responda com essas respostas".

A imagem abaixo mostra o primeiro dado da tabela expandido, para que possamos conferir como esse dado se comporta:

<br>

<div align="center">

<img src="https://iili.io/JhxSitp.png">

</div>

<br>

Perceba que, na imagem acima, possuímos uma **entrada** ("Sugira um slogan para uma campanha de reciclagem.\n") e uma **resposta esperada para essa entrada** ("1. \"Reduza, reutilize, recicle: Juntos por um futuro mais verde.\"\n2. \"Recicle hoje, para um amanhã melhor.\"\n3. \"Transforme seu lixo em tesouro - Recicle!\"\n4. \"Recicle para o ciclo da vida.\"\n5. \"Economize recursos, recicle mais.\"").


Ou seja, entregamos entradas ao modelo, e ele recebe as respostas esperadas (para cada entrada) como gabarito. Assim, esperamos que o modelo aprenda a nos responder utilizando as respostas que estão no rótulo (ou gabarito).

No exemplo acima, portanto, o treinamento é rotulado ou supervisionado por humanos. A maioria dos modelos de NLP atuais funcionam dessa forma.

Já os treinamentos **não supervisionados não possuem rótulos**. A IA aprende automaticamente os padrões necessários para acertar uma previsão. Esse tipo de treinamento pode ter resultados bem diferentes em relação ao treinamento supervisionado.

Para treinar, o modelo visita par a par (de perguntas e respostas). O ajuste dos parâmetros pode ocorrer a cada visita a um par, ou pode ser configurado para ser ajustado de dez em dez visitas aos pares de perguntas e respostas, por exemplo. Esse ajuste é feito por um hiperparâmetro que recebe o nome de **batch (ou lote)**. Você encontrará mais informações sobre o batch logo abaixo.

É importante ressaltar que é comum o modelo não utilizar todo o conjunto de dados (que contém os pares de perguntas e respostas) para treinar. Isso ocorre por conta da necessidade de validar a performance do modelo após o treinamento.

Ou seja, geralmente dividimos o conjunto de dados em **dados de treino** e **dados de teste**. Os dados de treino são utilizados para de fato treinar o modelo. Os dados de teste são usados para testar o modelo e checar se o treinamento foi eficiente.

Quando vamos treinar uma IA, definimos **hiperparâmetros** que vão descrever como o treinamento ocorrerá. Os principais são:

- **Epochs**: quantas vezes o modelo irá percorrer TODO o conjunto de dados de treino durante o treinamento;
- **Batches (lotes)**: quantos "pacotes" de dados serão utilizados (por exemplo: se tenho 1000 pares de perguntas e respostas e quero separar lotes (pacotes) de 100 pares, terei 10 batches ou lotes com 100 pares cada). O modelo ajustará seus parâmetros no final do processamento de cada batch. No exemplo, como temos 10 batches, em uma epoch o modelo ajusta os parâmetros 10 vezes. Isso serve para agilizar e organizar o treinamento;
- **Batch size**: quantas amostras temos em cada batch ou lote. No exemplo dado acima, teríamos um _batch size_ de 100.

#### **Problemas comuns no treinamento de modelos de IA**

Os modelos de IA também podem sofrer com problemas no treinamento. É comum vermos modelos que dão respostas imprecisas, como um modelo de reconhecimento facial que não te reconhece, um modelo de NLP que não te responde corretamente, etc.

Isso ocorre por conta de limitações nas redes neurais ou **problemas nos dados de treino**. Vamos investigar mais a fundo os motivos que levam a esse desajuste em um treinamento.

<br>
<br>

#### **Overfitting**

<br>

O **overfitting** é o ajuste exagerado dos parâmetros aos dados de treino. Ou seja, **o modelo fica viciado** em fornecer exatamente as respostas aos quais foram treinados.

Esse problema ocorre, na maioria das vezes, por haver **poucos dados** de treino. Também pode ocorrer por se utilizar um modelo muito complexo para a resolução de problemas mais simples, etc.

> Pense: se uma pessoa tivesse o objetivo, por exemplo, de aprender uma língua nova, e estudasse por apenas 10 dias. Em um teste de fluência, você acha que essa pessoa performaria bem? Isso é muito pouco provável, uma vez que um teste de fluência exige o domínio de muitas técnicas, como: domínio da fala, escrita e escuta, domínio de vocabulário, etc. Essa pessoa, estudando apenas 10 dias, tentaria resolver a prova usando o que sabe. Porém, em apenas 10 dias, essa pessoa não teve contato suficiente para expandir seu vocabulário, e nem para treinar as técnicas fundamentais da linguagem. Sendo assim, ficaria "viciada" em apenas usar o que sabe para tentar responder qualquer pergunta do teste, performando mal.
>
> Nesse exemplo, a pessoa foi treinada com pouca quantidade de técnicas, e acabou ficando "viciada" em fazer apenas o que sabe, perdendo desempenho no seu objetivo.
>
> Analogamente, caso existam poucos dados, um modelo de NLP pode ficar viciado em seus dados de treino e perder desempenho em seu objetivo.

Em modelos de NLP, esse problema se traduz da seguinte forma: suponha que você treine um modelo com perguntas e respostas referentes à noções de primeiros socorros, tendo esse caso como exemplo:

<br>

**_Pergunta (no dado de treino): "O que fazer em caso de queimadura de segundo grau?"_**

Resposta esperada (no dado de treino): _"Em caso de queimadura de segundo grau, a área afetada deve ser resfriada com água fria corrente por pelo menos 10 a 15 minutos, para reduzir a temperatura da pele e aliviar a dor. Em seguida, a área deve ser coberta com um curativo limpo e seco e o paciente deve procurar atendimento médico imediatamente."_

<br>

Então, você faz a seguinte pergunta ao modelo:

<br>

**_Pergunta: Se eu tivesse uma queimadura de segundo grau, quais seriam meus sintomas?_**

Resposta do modelo com overfitting: _"Em caso de queimadura de segundo grau, a área afetada deve ser resfriada com água fria corrente por pelo menos 10 a 15 minutos, para reduzir a temperatura da pele e aliviar a dor. Em seguida, a área deve ser coberta com um curativo limpo e seco e o paciente deve procurar atendimento médico imediatamente."_

Você percebe como o modelo responde "engessado" aos dados de treino? Isso indica que o modelo está viciado nos dados de treino e, consequentemente, possui dificuldade em generalizar seu conhecimento para responder outras perguntas. Esse seria um caso de **overfitting**.

<br>
<br>

#### **Underfitting**

<br>

O **underfitting** é o contrário do overfitting. Ou seja, os parâmetros não conseguem se ajustar bem aos dados de treino.

Isso pode acontecer por **falta de variedade nos dados de treino**, modelos muito simples para resolver problemas muito complexos, etc.

Enquanto o overfitting resulta em um modelo que só sabe responder de acordo com o que foi treinado (possui dificuldade em generalizar seu conhecimento), o underfitting resulta em um modelo que **não consegue responder de acordo com o que foi treinado**, gerando respostas equivocadas. Dessa forma, o modelo **alucina**, muitas vezes "inventando" informações para responder uma solicitação do usuário.

Veja, no exemplo abaixo, como um modelo de NLP com underfitting se comportaria:

**_Pergunta (no dado de treino): "O que fazer em caso de queimadura de segundo grau?"_**

Resposta esperada (no dado de treino): _"Em caso de queimadura de segundo grau, a área afetada deve ser resfriada com água fria corrente por pelo menos 10 a 15 minutos, para reduzir a temperatura da pele e aliviar a dor. Em seguida, a área deve ser coberta com um curativo limpo e seco e o paciente deve procurar atendimento médico imediatamente."_

<br>

Então, você faz a seguinte pergunta ao modelo:

<br>

**_Pergunta: Se eu tivesse uma queimadura de segundo grau, o que eu deveria fazer?_**

Resposta do modelo com overfitting: _"Em caso de queimadura de segundo grau, você deve estancar o sangramento, verificar o pulso da vítima e, caso não haja resposta imunológica, realizar o RCP."_

<br>

Perceba: a pergunta foi respondida baseada nos dados de treino, mas o modelo não foi capaz de capturar o padrão que fosse capaz de responder corretamente a pergunta, apresentando alucinações.

A imagem abaixo compara o underfitting, o overfitting e o fit adequado:

## **O fluxo de NLP**
### _Entenda como a informação caminha em modelos LLMs_

<br>

Quando enviamos uma prompt, por exemplo, ao **_Chat GPT_**, ele nos responde nos dando uma resposta coerente na maioria das vezes, certo?

Para que isso aconteça, nossa prompt precisa ser processada antes de entrar no modelo, como vimos nas sessões anteriores! O fluxo da informação será mais detalhado abaixo:

<div align="center">

`Entrada -> Tokenização -> Word Embedding -> Modelo -> Saída de texto`

</div>

1. A **entrada (prompt ou input**) é um texto gerado pelo usuário, que é enviado ao tokenizador;
2. **Tokenização:** os elementos do texto são quebrados em tokens e convertidos em números compreensíveis pelo modelo (input IDs);
3. **Word Embedding:** os tokens são convertidos em vetores semânticos. Esse é o processo que atribui significado a cada token;
4. Os vetores são enviados ao modelos para serem processados por Encoders, Decoders ou ambos (não se preocupe quanto a esses termos agora);
5. É gerada uma saída (_output_) em resposta à entrada (_input_).

Cabe ressaltar que, a depender do modelo, mais etapas podem se fazer necessárias. Serão exploradas mais informações sobre algumas dessas etapas nas próximas seções.

<br>

---

<br>

## **O que é a arquitetura Transformers?**
### _Compreenda o conceito do mecanismo de atenção dos LLMs_

<br>

Ao trabalharmos com LLMs, quase sempre estamos lidando com a arquitetura de modelos **Transformers**.

> A arquitetura de um modelo de IA é a forma em que uma ou diversas redes neurais de um modelo estão organizadas. Quando o modelo tem mais de uma rede neural, além de definir como as camadas de neurônios de cada rede estão organizadas, é importante definir como as diferentes redes estão interconectadas.

<div align="center">

<br>

<img src="https://miro.medium.com/v2/resize:fit:904/1*GAQrbFIV-G5cT3-OchMEHg.png" width=600px>

</div>

<br>

A arquitetura Transformers surgiu em 2017 e revolucionou o mundo das IAs generativas, principalmente, por conter um mecanismo de atenção muito sofisticado.

Podemos listar os seguintes pontos como vantagens de um modelo que utiliza a arquitetura Transformers:

- Ao invés de processar um texto palavra por palavra, o modelo é capaz de **processar todos os tokens da entrada simultaneamente**, trazendo velocidade de processamento;
- **Entendimento de relações muito mais complexas** entre as palavras;
- **Maior flexibilidade**, pois os modelos Transformers se adaptam muito bem à muitas das diferentes tarefas de NLP que vimos anteriormente;
- **Maior eficiência** no processamento, pois permite um processamento em paralelo. Para exemplificar melhor, pense numa fila de supermercado com 20 pessoas. Se existe apenas um caixa atendendo, o processamento demora muito. Se existem 10 caixas atendendo, o processamento é bem mais rápido, pois o conjunto de caixas atende diversos clientes de uma vez, caracterizando um processamento em paralelo. Isso acontece com as frases da entrada de um modelo Transformers.
- **Altíssima performance em modelos generativos**, como o ChatGPT ou modelos de tradução.

Os modelos com arquitetura Transformers podem possuir Encoders, Decoders ou ambos. Estes são blocos com diversas redes neurais dentro. Essas redes neurais possuem elementos que tornam a arquitetura muito especial: **camadas de atenção** que possibilitam o funcionamento do **mecanismo de atenção ou _self-attention_**.

Antes de entendermos o que é o mecanismo de atenção, é importante lembrarmos que **o significado de uma palavra está intimamente ligado ao seu contexto**. Isso significa que, para responder a uma input com alta eficácia, o modelo de IA deve ser capaz de entender como uma palavra pode ser ligada a outra dentro de uma frase.

Considere a seguinte frase (em inglês): "**_The animal didn't cross the street because it was too tired._**". Na frase, sabemos que o pronome "it" se refere a "The animal", mas para explicar essa relação ao modelo de IA, precisamos de artifícios como o mecanismo de atenção.

<div align="center">

<br>

![Mecanismo de atenção](https://jalammar.github.io/images/t/transformer_self-attention_visualization.png)

<br>

</div>

Para cada token da frase, o mecanismo de atenção dá um peso diferente para as outras palavras da frase. Suponhamos que o mecanismo esteja processando a palavra "it" (como na imagem acima). Para cada palavra restante na frase, o mecanismo atribuirá um peso que indica a importância de cada palavra em relação à palavra "it" (observe as linhas mais grossas, com peso maior).

Quanto maior o peso entre as palavras, maior a relação sintática dessa palavra com a palavra "it". Essa atribuição de peso é feita por diversas camadas de atenção dentro do modelo, através de processos complexos.

> **Mas como esse mecanismo pode ajudar o modelo a gerar boas respostas?**
>
> Pense: se o significado de uma palavra está relacionado ao contexto em que ela aparece, um modelo que entende melhor o contexto acaba entendendo melhor o significado de cada palavra naquela frase em que ela pertence.
> Dessa forma, o mecanismo de atenção dá ao modelo a habilidade de aprender relações muito mais complexas entre as palavras. Isso diminui, e muito, as chances de alucinação e problemas de concordância gramatical.

Assim, esse mecanismo ajuda o modelo a entender a relação das palavras com o contexto ao seu redor, gerando respostas muito mais precisas sintaticamente e semanticamente. Além disso, esse processo acontece ao mesmo tempo para todos os tokens da entrada (muito mais rápido que outras arquiteturas), sendo essa mais uma vantagem dos modelos com arquitetura Transformers.

A maioria dos LLMs, hoje em dia, utilizam essa arquitetura disruptiva.

## **Tokenização**
### _Preparando a entrada para o modelo_

<br>

Os modelos de NLP são formados por redes neurais artificiais, certo? Sendo assim, eles precisam de entradas numéricas para funcionarem. Porém, ao usarmos um modelo linguístico, as entradas são frases ou palavras que desejamos que o modelo interaja.

Por conta desse problema, existe o processo de **tokenização**. Esse processo separa uma entrada em pequenos pedaços (tokens). Em seguida, o modelo utiliza um algoritmo de codificação para transformar esses tokens em **IDs de entrada**, que nada mais são do que números que representam esses tokens. A ferramenta que faz a tokenização é denominado **tokenizador**. Comumente, tokenizadores também são capazes de realizarem a codificação dos tokens em input IDs, unindo todos os processos em uma ferramenta.

Na figura abaixo, a entrada "This is a input text." é tokenizada (quebrada em tokens) na primeira linha. Depois disso, na segunda linha, cada token é transformado em seu ID de entrada (_input ID_) correspondente.

<div align="center">

<br>

![Tokenização](https://iili.io/JvtTwR1.png)

</div>

<br>

A tokenização pode ocorrer de várias formas diferentes. O tokenizador pode separar a entrada por regras de pontuação, por espaçamento, etc. Algumas das formas de tokenização podem ser vistas na figura abaixo:

<div align="center">

<br>

![Token](https://miro.medium.com/v2/resize:fit:1400/1*CdjbU3J5BYuIi-4WbWnKng.png)

</div>

<br>

**Mas como o tokenizador define o número que cada palavra representa?**

Quando treinamos um modelo, ele aprende várias palavras. O conjunto de palavras que o modelo conhece é chamado de **vocabulário**.  Vamos supor que um modelo qualquer X conheça a palavra "ANEEL". Essa palavra vai ter uma posição no vocabulário. Vamos supor que "ANEEL" ocupe a posição 31887. O tokenizador, se receber a palavra "ANEEL", vai transformá-la no número 31887.

**Mas e se a palavra não estiver no vocabulário do modelo?**

Os modelos atuais possuem um token separado apenas para palavras desconhecidas. Assim, quando uma palavra desconhecida é processada pelo tokenizador, ele a transforma no token que corresponde a palavras desconhecidas.

<br>

Dessa forma, o texto é transformado em uma série de tokens pelo tokenizador. Esses tokens serão enviados para o mecanismo de _Word Embedding_.

## **Word Embedding**
### _Como os modelos de NLP aprendem os significados das palavras?_

<br>

Para entender o significado das palavras, os modelos utilizam o que chamamos de **word embedding**, ou incorporação de palavras.

Como vimos, o modelo analisa a relação das palavras com o seu contexto (passado e futuro) através das camadas de atenção. Isso acontece pois o **significado** de uma palavra está intimamente relacionado ao contexto em que ela está inserida.

Vamos pensar na palavra "rainha". Em termos de significado, ela está mais próxima da palavra "rei" ou da palavra "abacate"?

Para entender como o computador responde a essa pergunta, vamos um pouco mais a fundo.

Considere as seguintes características: realeza e gênero. Essas características serão chamadas de dimensões. Podemos agora dar uma pontuação entre -1 e 1 para a palavra "rainha" e para a palavra "rei" em relação a cada uma dessas dimensões.

<br>

> Para a dimensão "realeza", -1 significa um plebeu, e 1 significa um alto nobre.
>
> Para a dimensão "gênero", -1 significa feminino e 1 significa masculino.

<br>

É seguro dizer que a palavra "rainha" teria um valor próximo de 1 para realeza, e próximo de -1 para gênero. Já a palavra "rei" teria um valor próximo de 1 para realeza, e próximo de 1 para gênero.  Percebe que estamos descrevendo o significado dessas palavras em relação a essas características?

Dessa forma, podemos descrever uma palavra em relação à várias dimensões (características) diferentes, sendo que cada dimensão teria um número entre -1 e 1 para descrever a palavra em relação àquela característica.

Naturalmente, quanto mais dimensões tivermos para descrever as palavras, melhor será a compreensão dessas palavras. Além disso, a distinção entre as palavras também será melhor compreendida pelo modelo. Você entenderá melhor no que se segue.

Um vetor é uma coleção de dados. Podemos ter vetores que carregam números, palavras, caracteres, etc. Agora imagine que criamos um vetor para a palavra "Rainha", na intenção de guardar os números entre -1 e 1 que a descrevem, para cada dimensão. Dessa forma, estamos construindo um vetor que **guarda o significado dessa palavra**.

Para que fique mais claro: na imagem abaixo, foram construídos os vetores de "Rainha" e de "Rei" em relação às dimensões (Gênero, Realeza, Fruta, Violência, etc.).

<br>

<div align="center">
  <img src="https://miro.medium.com/v2/resize:fit:640/format:webp/0*DyatUMUNFLSdogD3">
  <img src="https://miro.medium.com/v2/resize:fit:518/format:webp/0*J0EcwwgwYTE5aBp2">
</div>

<br>

Para esse exemplo, vamos inicialmente usar apenas duas dimensões: Realeza e Gênero. Agora imagine que cada uma dessas dimensões forme um eixo em um gráfico cartesiano (Realeza no eixo X e Gênero no eixo Y, por exemplo). Nesse espaço de duas dimensões, podemos colocar um ponto que represente a posição da palavra "rainha" e outro para a palavra "rei", seguindo suas pontuações nas dimensões realeza e gênero. Essa representação é chamada de representação espacial de duas dimensões. A distância entre os pontos é uma distância que chamamos de distância vetorial.

<br>

<div align="center">
<img src="https://iili.io/JVnz34p.jpg">
</div>

<br>

Na imagem acima, quanto mais próximas estão as palavras, mais parecidas elas são em termos de significado.

Imagine agora que façamos esse processo para a palavra "Abacate" (suponhamos que o mesmo assumiria um valor próximo a 0.5 para Gênero e um valor próximo a -0.5 para Realeza). Se o colocássemos no gráfico, veríamos que, em termos espaciais, a palavra "Abacate" está está quase tão próxima de "Rainha" quanto de "Rei", conforma consta na imagem abaixo:

<br>

<div align="center">
<img src="https://iili.io/JvDq8tR.png">
</div>

<br>

Mas "Rei" e "Rainha" deveriam estar próximos e  a palavra "Abacate" deveria estar mais afastada de ambas, certo? Isso não aconteceu no nosso exemplo, pois possuímos apenas 2 dimensões em nossa análise (eixos x e y na imagem).

Por esse motivo, precisamos de **mais dimensões para descrever** bem uma palavra. Com um bom número de dimensões, a distância vetorial entre "rei" e "rainha" será bem menor que a distância entre "abacate" e rainha.

<br>

> **Curiosidade**: um modelo moderno geralmente define mais de 100 dimensões para pontuar as palavras. Assim, não conseguimos imaginar a posição dessas palavras visual ou espacialmente, apenas matematicamente.

<br>

Voltando ao computador, podemos dizer, de forma bem resumida, que ele define dimensões, pontua cada token para essas dimensões e define uma posição espacial para cada token baseada nessas pontuações, tudo sendo *baseado no contexto* ao redor das palavras.

Com as posições de cada palavra, o computador consegue calcular a distância vetorial entre elas.

As palavras em posições mais próximas (em termos vetoriais ou espaciais) são consideradas mais próximas, *em significado*, pelo computador. Dessa forma, o computador consegue "entender" o significado das palavras e utilizá-las na circunstância correta.

O _word embedding_ é o processo que transforma cada palavra em um vetor (array) de significado (cheio de pontuações em relação às dimensões), constituindo um "endereço" espacial para essa palavra. Fazendo isso para várias palavras, o processo ajuda o computador a entender a proximidade dos significados entre as diferentes palavras de uma entrada.

Veja, na imagem abaixo, que cada input ID foi transformado em um vetor de pontuações.

</div>

Dessa forma, podemos concluir que, ao utilizar esse mecanismo nos tokens, o modelo está pronto para processar essa entrada em suas camadas de redes neurais artificiais, como vimos nas seções anteriores, e, assim, gerar uma saída de valor para o usuário.
